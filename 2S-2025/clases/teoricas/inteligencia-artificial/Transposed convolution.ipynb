```python
import torch
import torch.nn as nn

#Ejemplo 1
# With square kernels and equal stride
m = nn.ConvTranspose2d(16, 33, 3, stride=2)
input = torch.randn(20, 16, 50, 100)

output = m(input)
print(output.size())

#Ejemplo 2
# exact output size can be also specified as an argument
input = torch.randn(1, 16, 12, 12)
downsample = nn.Conv2d(16, 16, 4, stride=2, padding=1)
upsample = nn.ConvTranspose2d(16, 16, 4, stride=2, padding=1)
h = downsample(input)

print(h.size())
output = upsample(h)

print(output.size())
```

1 - Probar con kernels asimetricos:
  m = nn.ConvTranspose2d(16, 33, (3, 5), stride=(2, 1), padding=(4, 2))
  
2 - ¿Qué pasa si el kernel es de 3 en el segundo ejemplo?, se puede explicitar el tamaño de salida:
upsample(h, output_size=input.size())
